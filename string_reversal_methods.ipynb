# Reverse a String in Python

This notebook demonstrates various methods to reverse a string in Python, including:
- String slicing
- Loop-based reversal
- Recursive reversal
- Performance comparison

We'll explore each method and benchmark them to determine which approach is most efficient.

## Section 1: Import Required Libraries

We'll import the `time` module to measure the execution time of different string reversal methods.

```python
import time
```

## Section 2: Reverse a String Using Slicing

Python's string slicing with a negative step is one of the simplest and most efficient ways to reverse a string.

**Syntax:** `string[::-1]`

This works by:
- Starting from the end of the string (no start index specified means start from the end with step -1)
- Moving backwards through each character
- Returning the reversed string

```python
# Method 1: String Slicing
def reverse_string_slicing(s):
    """
    Reverse a string using Python's slicing with negative step.
    
    Args:
        s (str): The string to reverse
        
    Returns:
        str: The reversed string
    """
    return s[::-1]

# Test the slicing method
test_string = "Hello, World!"
reversed_slicing = reverse_string_slicing(test_string)
print(f"Original string: {test_string}")
print(f"Reversed (slicing): {reversed_slicing}")
```

## Section 3: Reverse a String Using a Loop

This method uses a for loop to iterate through the string in reverse order by accessing characters using negative indices.

```python
# Method 2: Using a Loop
def reverse_string_loop(s):
    """
    Reverse a string using a for loop.
    
    Args:
        s (str): The string to reverse
        
    Returns:
        str: The reversed string
    """
    reversed_str = ""
    for i in range(len(s) - 1, -1, -1):
        reversed_str += s[i]
    return reversed_str

# Test the loop method
reversed_loop = reverse_string_loop(test_string)
print(f"Original string: {test_string}")
print(f"Reversed (loop): {reversed_loop}")
```

## Section 4: Reverse a String Using Recursion

A recursive approach breaks the string reversal problem into smaller subproblems. The base case is when the string is empty or has one character, and the recursive case removes the first character and concatenates it at the end of the reversed substring.

```python
# Method 3: Using Recursion
def reverse_string_recursion(s):
    """
    Reverse a string using recursion.
    
    Args:
        s (str): The string to reverse
        
    Returns:
        str: The reversed string
    """
    # Base case: empty string or single character
    if len(s) == 0:
        return s
    # Recursive case: reverse the rest and add first character at the end
    else:
        return reverse_string_recursion(s[1:]) + s[0]

# Test the recursion method
reversed_recursion = reverse_string_recursion(test_string)
print(f"Original string: {test_string}")
print(f"Reversed (recursion): {reversed_recursion}")
```

## Section 5: Compare Performance of Different Methods

Now let's benchmark all three methods to see which one is most efficient. We'll test them with strings of varying lengths and measure execution time.

```python
# Performance comparison
def benchmark_methods(string_length):
    """
    Benchmark the three string reversal methods.
    
    Args:
        string_length (int): The length of the test string
        
    Returns:
        dict: Dictionary containing execution times for each method
    """
    # Create a test string
    test_str = "A" * string_length
    
    results = {}
    
    # Benchmark slicing method
    start = time.time()
    for _ in range(100000):
        reverse_string_slicing(test_str)
    results['Slicing'] = time.time() - start
    
    # Benchmark loop method
    start = time.time()
    for _ in range(100000):
        reverse_string_loop(test_str)
    results['Loop'] = time.time() - start
    
    # Benchmark recursion method (with smaller iterations due to overhead)
    start = time.time()
    for _ in range(10000):
        reverse_string_recursion(test_str)
    results['Recursion'] = time.time() - start
    
    return results

# Test with different string lengths
print("Performance Comparison (100,000 iterations for slicing/loop, 10,000 for recursion)")
print("=" * 70)

for length in [10, 50, 100]:
    print(f"\nString length: {length} characters")
    print("-" * 70)
    timings = benchmark_methods(length)
    for method, elapsed_time in timings.items():
        print(f"{method:15} | Time: {elapsed_time:.6f} seconds")
```

## Performance Analysis Summary

Based on the benchmarks:

1. **Slicing Method** — The fastest approach! It's implemented in C at the CPython level, making it highly optimized.

2. **Loop Method** — Slower than slicing but still reasonable. String concatenation in the loop creates new string objects, which adds overhead.

3. **Recursion Method** — The slowest method due to function call overhead and repeated string slicing in the recursive calls. It's also not ideal for very long strings due to recursion depth limits.

### Recommendation

**Use the slicing method (`s[::-1]`) for production code** because it:
- Is the most efficient
- Is the most Pythonic
- Is readable and concise
- Handles edge cases elegantly

The loop and recursion methods are useful for **educational purposes** to understand different algorithmic approaches.

```python
# Bonus: Alternative approaches

# Method 4: Using reversed() built-in function
def reverse_string_reversed_builtin(s):
    """
    Reverse a string using the built-in reversed() function.
    
    Args:
        s (str): The string to reverse
        
    Returns:
        str: The reversed string
    """
    return ''.join(reversed(s))

# Test it
reversed_builtin = reverse_string_reversed_builtin(test_string)
print(f"Original string: {test_string}")
print(f"Reversed (reversed() builtin): {reversed_builtin}")
print("\nAll methods produce the same result:", 
      reversed_slicing == reversed_loop == reversed_recursion == reversed_builtin)
```
